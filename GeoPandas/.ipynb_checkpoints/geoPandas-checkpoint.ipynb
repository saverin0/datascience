### Set operations on geometries 

```python
circle_1 = Point(0,0).buffer(2)
circle_2 = Point(3,0).buffer(2)
circle_1
```

```python
circle_1.union(circle_2)
```

```python
circle_1.intersection(circle_2)
```

```python
circle_1.difference(circle_2)
```

```python
circle_2.difference(circle_1)
```

```python
point = Point(0,0)
triangle_1 = Polygon([(-1,-1),(0,1),(1,-1)])
triangle_2 = Polygon([(3,4),(5,14),(10,10)])
triangle_1
```

```python
triangle_2
```

```python
point.within(triangle_1)
```

```python
point.within(triangle_2)
```

```python
point.union(triangle_1).union(triangle_2)
```

```python
point.distance(triangle_1)
```

```python
point.distance(triangle_2)
```

```python
#let's create a car 
```

```python
top = Polygon([(1.5,1),(3.5,1),(3.5,2),(1.5,2)])
top
```

```python
bottom = Polygon([(0,0),(5,0),(5,1),(0,1)])
bottom
```

```python
wheel1 = Point(1,-0.5).buffer(0.5)
wheel1
```

```python
wheel2 = Point(4,-0.5).buffer(0.5)
wheel2
```

```python
from shapely.ops import unary_union
```

```python
car = top.union(bottom).union(wheel1).union(wheel2)
car
```

```python
car1 = unary_union([bottom, top, wheel1, wheel2])
car1
```

```python
import matplotlib.pyplot as plt
```

```python
fig, ax = plt.subplots(figsize=(6, 3))
for geom in [bottom, top, wheel1, wheel2]:
    x, y = geom.exterior.xy
    ax.fill(x, y, alpha=0.7)

ax.set_aspect('equal')
plt.title("Car Shape Using Shapely")
plt.grid(True)
plt.show()
```
