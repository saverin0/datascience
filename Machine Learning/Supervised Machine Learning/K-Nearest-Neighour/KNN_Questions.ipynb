1. Is KNN a method? Why is KNN not a learning algorithm?
Answer: KNN is not a learning algorithm because it does not involve a training phase where the model learns parameters from the data. Instead, it is a lazy learning algorithm that stores the entire dataset and makes predictions during inference by finding the nearest neighbors.
2. Can we use KNN for handling missing values?
Answer: Yes, KNN can be used for handling missing values. By finding the nearest neighbors of a data point with missing values, we can impute the missing values using the mean, median, or mode of the corresponding feature values from the neighbors.
3. Can we give KNN a large dataset or a small dataset?
Answer: KNN works better with small datasets because it requires storing the entire dataset and performing distance calculations for each prediction. For large datasets, KNN can become computationally expensive and memory-intensive.
4. What is the value of k (nearest neighbors)?
Answer:
k is the number of nearest neighbors considered for making predictions.
How do we decide k?: k is chosen based on cross-validation to optimize accuracy.
Odd or even k? Why?: k is usually chosen as an odd number to avoid ties in classification tasks.
Effect on accuracy: A small k can lead to overfitting, while a large k can lead to underfitting.
5. How do we conclude the value of k?
Answer:
k is a hyperparameter that can be tuned using techniques like grid search or cross-validation.
The value of k is evaluated by measuring the model's accuracy on a validation set.
6. KNN: Regression or Classification?
Answer: KNN can be used for both regression and classification tasks:
Regression: The prediction is the average (or weighted average) of the target values of the k nearest neighbors.
Classification: The prediction is the majority class among the k nearest neighbors.
7. What is the difference between lazy learners and eager learners? Why is KNN a lazy learner?
Answer:
Lazy learners: Do not build a model during training. They store the dataset and make predictions during inference. KNN is a lazy learner because it does not involve a training phase.
Eager learners: Build a model during training and use it for predictions. Examples include decision trees and neural networks.
KNN is a lazy learner because it defers computation (distance calculation and prediction) until the query time.
